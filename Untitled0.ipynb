## Simple Python Code for GitHub Assignment

This code defines a function to add two numbers and then demonstrates how to use it. It's a basic example that can be easily understood and uploaded to GitHub.

```python
def add_numbers(a, b):
    """This function takes two numbers and returns their sum."""
    return a + b

# Demonstrate the function
num1 = 5
num2 = 10
sum_result = add_numbers(num1, num2)

print(f"The sum of {num1} and {num2} is: {sum_result}")

# You can also try with different numbers
print(f"The sum of 7 and 3 is: {add_numbers(7, 3)}")
```
